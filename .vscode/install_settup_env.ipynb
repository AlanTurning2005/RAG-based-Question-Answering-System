```python
import sys
import io

# sys.stdout = io.TextIOWrapper(sys.stdout.buffer, encoding='utf-8')

import sentence_transformers
import faiss
import PyPDF2

print("Môi trường RAG đã sẵn sàng hoạt động tuyệt vời!")
print(f"Phiên bản Sentence-Transformers: {sentence_transformers.__version__}")
from sentence_transformers import SentenceTransformer
import numpy as np
import torch 
import torch.nn.functional as F
```

```python
def cosine_similarity(v1, v2):
    dot = np.dot(v1, v2)
    norm = np.linalg.norm(v1)* np.linalg.norm(v2)
    return dot / norm if norm != 0 else 0.0


if __name__ == "__main__":
    # v1 = np.array([1, 2, 3])
    # v2 = np.array([4, 5, 6])
    # similarity = cosine_similarity(v1, v2)
    # print(f"Cosine similarity between {v1} and {v2} is: {similarity}")
    model  = SentenceTransformer("all-MiniLM-L6-v2")
    sentences = [' the cat is running', "the weather is bad", "we can run on the field",
                "do you love me forever", "i really love you my son", "this book is so boring", 
                " Watch out !", "I am Steve", "The dogs are hungry", "don't go to swim today !"]
    embeddings = model.encode(sentences)
    print(embeddings.shape)
    embed_tensor = torch.tensor(embeddings)
    normalize_v = F.normalize(embed_tensor, dim = 1)
    cosine_sim_matrix  = torch.mm(normalize_v, normalize_v.t())
    print(f"Do tuong dong cua 10 cau sau khi embedding sentences : {cosine_sim_matrix}")
    
```

### Data Analysis 

**Thống kê dataset:** Chia thành RACE-M (THCS: 25,421 câu hỏi) và RACE-H (THPT: 62,445 câu hỏi).Bài đọc của RACE-H dài hơn (trung bình 353.1 từ) và từ vựng phong phú hơn (125,120 từ) so với RACE-M
Phân loại mức độ suy luận (Reasoning Types):
    Word matching (Khớp từ khóa): 15.8%
    Paraphrasing (Diễn giải lại): 19.2%
    Single-sentence reasoning (Suy luận đơn câu): 33.4%
    Multi-sentence reasoning (Suy luận đa câu): 25.8%
    Insufficient/Ambiguous (Mơ hồ): 5.8%

Tổng tỷ lệ câu hỏi cần suy luận đạt 59.2%, cao hơn vượt trội so với CNN (21%) hay SQuAD (20.5%).

Phân loại chi tiết dạng câu hỏi: Bao gồm suy luận chi tiết (detail reasoning), hiểu toàn cảnh (whole-picture reasoning), tóm tắt bài đọc (passage summarization), phân tích thái độ tác giả (attitude analysis) và tri thức thế giới (world knowledge)

### 2. CHUNKING
- **StackOverflow Blog: Breaking up is hard to do** [https://stackoverflow.blog/2024/12/27/breaking-up-is-hard-to-do-chunking-in-rag-applications/]

    **2.1. Why is chunking important?**

        + Chunking quá lớn -> mất khả năng cụ thể 

        + chunking quá ít -> mất bối cảnh.

    **=> Kích thước phải phù hợp**

        + kích thước của chunk quan trọng, ngoài ra kích thước của query vector cần phải matching với chunking vector. 

        + metadata là 1 cách để liên kết tới phần gốc của văn bản.

    **2.2 Chunking strategies**
        + Cách cơ bản : chia theo fixed-size. (phương pháp rẻ nhất về lượng tính toán ta cần, nhưng nó không tính đến ngữ cảnh của nội dung đang chia nhỏ. )

        + random chunk sizes : gồm nhiều loại tài liệu, do nhiều bối cảnh, chủ đề (không nên làm do tính ngẫu nhiên )
    
        + sliding windows methods : đối với document thứ n+1 sẽ overlap lại document thứ n và obtain thêm document của chính n+1.

        + Context-aware chunking: chia dựa trên dấu câu (yêu cầu tính toán, **document thay đổi theo thời gian gặp struggle**)  

        + adaptive chunking: Dùng ML để chia thành các unit nhỏ hơn có context tốt, highly-tailored(computation cao)


    **Kết luận** : Khi bạn kiểm tra các phương pháp chia đôi, hãy kiểm tra kết quả của hệ thống RAG với các truy vấn mẫu. Đánh giá họ bằng đánh giá của con người và với các chuyên gia đánh giá LLM. Khi bạn đã xác định được phương pháp nào luôn hiệu quả hơn, bạn có thể nâng cao kết quả bằng cách lọc kết quả dựa trên điểm tương đồng cosine.
